```python
import requests
```

```python
api_path="https://qvapay.com/api/p2p"
```

```python
def fetch_data_from_api(api_token=None):
    url = api_path 
    headers = {}
    if api_token:
        headers["Authorization"] = f"Bearer {api_token}"
    response = requests.get(url, headers=headers)
    if response.status_code == 200:
        return response.json()
    else:
        raise Exception(f"Error fetching data from API: {response.status_code}")

def main(api_token=None):
    try:
        data = fetch_data_from_api(api_token)
        #print("Offers Data:", data)
    except Exception as e:
        print(e)
    return(data)
```

```python
def obtener_last_page():
    data = main()
    last_page = data['last_page']
    return last_page
```

```python
num_paginas = obtener_last_page()
```

```python
datos =[]
for i in range(1, num_paginas+1):
    api_path = f"https://qvapay.com/api/p2p?page={i}"
    data = main()
    datos.extend(data['data'])
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
from ipywidgets import interact, Text

# Example dataset
filtered_base = pd.DataFrame({
    'user': ['Alice', 'Bob', 'Alice', 'Charlie', 'Bob', 'Alice', 'Bob', 'Alice', 'Charlie'],
    'coin': ['BTC', 'BTC', 'ETH', 'BTC', 'ETH', 'BTC', 'BTC', 'BTC', 'ETH'],
    'price': [50000, 51000, 1800, 50500, 1900, 49500, 49000, 50000, 1850],
    'amount': [0.1, 0.2, 1.0, 0.15, 2.0, 0.1, 0.3, 0.5, 0.2],
    'type': ['buy', 'sell', 'buy', 'sell', 'sell', 'buy', 'sell', 'buy', 'buy'],
    'date': pd.to_datetime(['2023-01-01', '2023-01-01', '2023-01-02', '2023-01-02', 
                            '2023-01-03', '2023-01-03', '2023-01-03', '2023-01-04', '2023-01-04'])
})

# Function to plot transactions for a specific user
def plot_user_transactions(user):
    # Filter data for the selected user
    user_data = filtered_base[filtered_base['user'] == user]
    
    if user_data.empty:
        print(f"No transactions found for user: {user}")
        return
    
    # Calculate cumulative stock
    user_data['cumulative_stock'] = user_data.apply(
        lambda row: row['amount'] if row['type'] == 'buy' else -row['amount'], axis=1
    ).cumsum()

    fig, ax = plt.subplots(figsize=(12, 6))
    
    # Get unique days
    unique_days = user_data['date'].dt.date.unique()
    
    # Draw day separators
    for idx, day in enumerate(unique_days):
        day_data = user_data[user_data['date'].dt.date == day]
        start = day_data.index[0] - 0.5
        end = day_data.index[-1] + 0.5
        color = 'lightgray' if idx % 2 == 0 else 'white'
        ax.add_patch(Rectangle((start, ax.get_ylim()[0]), end - start, ax.get_ylim()[1] - ax.get_ylim()[0], color=color, alpha=0.2))
    
    # Plot floating bars for each transaction
    prev_stock = 0  # Start cumulative stock at 0
    for i, row in user_data.iterrows():
        new_stock = prev_stock + (row['amount'] if row['type'] == 'buy' else -row['amount'])
        color = 'green' if row['type'] == 'buy' else 'gray'
        
        # Plot a bar from the previous stock to the new stock
        ax.bar(i, new_stock - prev_stock, bottom=prev_stock, color=color, label='Buy' if color == 'green' else 'Sell')
        
        prev_stock = new_stock  # Update the stock
    
    # Configure plot
    ax.set_xticks(user_data.index)
    ax.set_xticklabels(user_data['date'].dt.strftime('%H:%M:%S'), rotation=45)
    ax.set_ylabel('Stock')
    ax.set_xlabel('Transactions')
    ax.set_title(f'Transactions for User: {user}')
    ax.legend(['Buy', 'Sell'])
    plt.tight_layout()
    plt.show()

# Interactive widget to input user name
interact(plot_user_transactions, user=Text(value='Alice', description='User:'))
```
