```python
import pandas as pd

# Example dataset
data = {
    "Year": [2018, 2019, 2020, 2021, 2022],
    "Sales": [500, 600, 700, 800, 900],
    "Profit": [50, 60, 70, 80, 90],
    "Expenses": [67, 89, 64, 720, 810]
}
df = pd.DataFrame(data)
```

```python
import plotly.graph_objects as go

# Create the initial figure
fig = go.Figure()

# Add traces for each parameterizable field
fields = ["Sales", "Profit", "Expenses"]
for field in fields:
    fig.add_trace(go.Scatter(x=df["Year"], y=df[field], mode="lines+markers", name=field, visible=False))

# Make the first field (Sales) visible by default
fig.data[0].visible = True

# Define the dropdown menu
buttons = []
for i, field in enumerate(fields):
    buttons.append(dict(
        label=field,
        method="update",
        args=[
            {"visible": [j == i for j in range(len(fields))]},  # Toggle visibility
            {"title": f"{field} Over Time"}  # Update title
        ]
    ))

# Add the dropdown menu to the figure
fig.update_layout(
    updatemenus=[
        dict(
            active=0,
            buttons=buttons,
            direction="down",
            x=0.1,
            y=1.2,
            showactive=True
        )
    ],
    title="Sales Over Time",
    xaxis_title="Year",
    yaxis_title="Value"
)

# Show the figure
fig.show()
```
